```python
st = "adad"
# length     : len(string)
# upper_case : string.upper() ----> this needs to be stored
# lower_case : string.lower()
# title      : string.title() ----> first letter of every word to capital
# capitalize : string.capitalize() ----> only the very first letter to capital
# rem_spaces : string.strip() ----> removes spaces from start and end only

string = "   mayank rajput room number 38.  "
print(len(string))
print(string.upper())
print(string.title())
print(string.capitalize())
print(string.strip())
```
