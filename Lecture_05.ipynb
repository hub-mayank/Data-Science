# Question-1 (numpy)

```python
import numpy as np
a = np.random.randint(1,91,(5,5))
# b = a.reshape(5,5)
print(a)

m = np.mean(a,axis = 1)
print("\n Mean(row): ",m)
mn = np.mean(a,axis = 0)
print("\n Mean(column): ",mn)

s = np.sum(a,axis=1)
print("\n Sum(row) : ",s)
sm = np.sum(a,axis=0)
print("\n Sum(column) : ",sm)

a = np.sort(a,axis=1)
a = np.sort(a,axis=0)
print("\nSorted :\n",a)

mod = np.where(a%2==0,-1,a)
print("\nModified :\n",mod)


cnt = np.sum(np.where(mod<0,1,0))

print("\nNo. of negatives:",cnt)


```

# Question-2

```python
stock = np.random.randint(20,101,(5,5))
print("\nStock:\n",stock)
demand = np.random.randint(10,60,(5,5))
print("\nDemand:\n",demand)

stock = stock - demand
print("\nAfter Market :\n",stock)

stock = np.where(stock<0,0,stock)
print("\nCurrent stocks :\n",stock)


cnt = np.sum(np.where(stock<=0,1,0))
print("\nProducts with shortage", cnt)
print("\nProducts with no shortage :",25-cnt)

stock = np.sort(stock,axis=0)
print("\nSorted:\n",stock)





```
