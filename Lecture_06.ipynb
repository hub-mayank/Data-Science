**Rows & Columns {iloc & loc}**


iloc works only on indices and loc works on labels too


iloc makes indexing of it's own but loc searches for the specific index

```python
display(df2[['internet_usage', 'study_hours']])   #displaying specific rows
display(df2.study_hours) # -----> NOT recommended

df2.iloc[0:5]           #range of rows

df2.loc[df2['study_hours'] == 11, ['study_hours', 'sleep_hours']]
display(df2.loc[df2['study_hours'] == 11, ['study_hours', 'sleep_hours']].iloc[2:5])   #this is displaying the data with value of study_hours equal to 11 with columns - study_hours and sleep_hours, in which we are printing 2 to 5 rows (range)


```

**Assignment - 8**
